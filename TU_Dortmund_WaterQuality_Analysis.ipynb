```python
import pandas as pd

# Read the CSV file (it uses semicolons ';' to separate columns)
df = pd.read_csv('waterquality.csv', sep=';')

# Show the first 5 rows
df.head()
```

```python
# Check how many rows and columns exist
print("Rows and columns:", df.shape)

# Check for missing values
print(df.isnull().sum())

# Count how many measurements are in each section and season
pd.crosstab(df['river_section'], df['saison'])
```

```python
# 1. Oxygen by Season
print("OXYGEN BY SEASON:")
print("Mean:\n", df.groupby('saison')['oxygen_mgL'].mean().round(3))
print("SD:\n", df.groupby('saison')['oxygen_mgL'].std().round(3))
print("Median:\n", df.groupby('saison')['oxygen_mgL'].median().round(3))
print("IQR:\n", (df.groupby('saison')['oxygen_mgL'].quantile(0.75) - df.groupby('saison')['oxygen_mgL'].quantile(0.25)).round(3))
```

```python
# 2. Nitrate by Season
print("NITRATE BY SEASON:")
print("Mean:\n", df.groupby('saison')['nitrate_mgL'].mean().round(3))
print("SD:\n", df.groupby('saison')['nitrate_mgL'].std().round(3))
print("Median:\n", df.groupby('saison')['nitrate_mgL'].median().round(3))
print("IQR:\n", (df.groupby('saison')['nitrate_mgL'].quantile(0.75) - df.groupby('saison')['nitrate_mgL'].quantile(0.25)).round(3))
```

```python
# 3. Oxygen by River Section
print("OXYGEN BY SECTION:")
print("Mean:\n", df.groupby('river_section')['oxygen_mgL'].mean().round(3))
print("SD:\n", df.groupby('river_section')['oxygen_mgL'].std().round(3))
print("Median:\n", df.groupby('river_section')['oxygen_mgL'].median().round(3))
print("IQR:\n", (df.groupby('river_section')['oxygen_mgL'].quantile(0.75) - df.groupby('river_section')['oxygen_mgL'].quantile(0.25)).round(3))
```

```python
# 4. Nitrate by River Section
print("NITRATE BY SECTION:")
print("Mean:\n", df.groupby('river_section')['nitrate_mgL'].mean().round(3))
print("SD:\n", df.groupby('river_section')['nitrate_mgL'].std().round(3))
print("Median:\n", df.groupby('river_section')['nitrate_mgL'].median().round(3))
print("IQR:\n", (df.groupby('river_section')['nitrate_mgL'].quantile(0.75) - df.groupby('river_section')['nitrate_mgL'].quantile(0.25)).round(3))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Graph 1: Seasons
seasons = ['spring', 'summer', 'autumn', 'winter']
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

sns.boxplot(data=df, x='saison', y='oxygen_mgL', order=seasons, ax=ax[0])
ax[0].set_title('Dissolved Oxygen by Season')
ax[0].set_xlabel('Season')
ax[0].set_ylabel('Oxygen (mg/L)')

sns.boxplot(data=df, x='saison', y='nitrate_mgL', order=seasons, ax=ax[1])
ax[1].set_title('Nitrate by Season')
ax[1].set_xlabel('Season')
ax[1].set_ylabel('Nitrate (mg/L)')

plt.tight_layout()
plt.savefig('Figure1_Seasons.png', dpi=300)
plt.show()
```

```python
from scipy import stats

# Separate Spring and Autumn data
spring_oxy = df[df['saison'] == 'spring']['oxygen_mgL']
autumn_oxy = df[df['saison'] == 'autumn']['oxygen_mgL']

# Step A: Check assumptions for Oxygen
print("Shapiro Spring Oxygen:", stats.shapiro(spring_oxy))
print("Shapiro Autumn Oxygen:", stats.shapiro(autumn_oxy))
print("Levene Equal Variance Oxygen:", stats.levene(spring_oxy, autumn_oxy))

# Step B: Both p-values > 0.05 (normal & equal variance) -> Run Two-Sample t-test
print("t-test Oxygen:", stats.ttest_ind(spring_oxy, autumn_oxy, equal_var=True))
```

```python
spring_nit = df[df['saison'] == 'spring']['nitrate_mgL']
autumn_nit = df[df['saison'] == 'autumn']['nitrate_mgL']

# Step A: Check assumptions for Nitrate
print("Shapiro Spring Nitrate:", stats.shapiro(spring_nit))
print("Shapiro Autumn Nitrate:", stats.shapiro(autumn_nit))
print("Levene Equal Variance Nitrate:", stats.levene(spring_nit, autumn_nit))

# Step B: Shapiro p < 0.05 (not strictly normal) -> Run Mann-Whitney U test
print("Mann-Whitney Nitrate:", stats.mannwhitneyu(spring_nit, autumn_nit))
```

```python
a1_nit = df[df['river_section'] == 'A1_headwaters']['nitrate_mgL']
a2_nit = df[df['river_section'] == 'A2_upper_course']['nitrate_mgL']
a3_nit = df[df['river_section'] == 'A3_middle_course']['nitrate_mgL']
a4_nit = df[df['river_section'] == 'A4_lower_course']['nitrate_mgL']
a5_nit = df[df['river_section'] == 'A5_mouth']['nitrate_mgL']

# Check normality for each section
print("Shapiro A1:", stats.shapiro(a1_nit))
print("Shapiro A2:", stats.shapiro(a2_nit))
print("Shapiro A3:", stats.shapiro(a3_nit))
print("Shapiro A4:", stats.shapiro(a4_nit))
print("Shapiro A5:", stats.shapiro(a5_nit))

# Check equal variance across all 5 sections
print("Levene Nitrate:", stats.levene(a1_nit, a2_nit, a3_nit, a4_nit, a5_nit))

# All p > 0.05 -> Run One-Way ANOVA
print("ANOVA Nitrate:", stats.f_oneway(a1_nit, a2_nit, a3_nit, a4_nit, a5_nit))
```

```python
a1_oxy = df[df['river_section'] == 'A1_headwaters']['oxygen_mgL']
a2_oxy = df[df['river_section'] == 'A2_upper_course']['oxygen_mgL']
a3_oxy = df[df['river_section'] == 'A3_middle_course']['oxygen_mgL']
a4_oxy = df[df['river_section'] == 'A4_lower_course']['oxygen_mgL']
a5_oxy = df[df['river_section'] == 'A5_mouth']['oxygen_mgL']

# Check normality for each section
print("Shapiro A1:", stats.shapiro(a1_oxy))
print("Shapiro A2:", stats.shapiro(a2_oxy))
print("Shapiro A3:", stats.shapiro(a3_oxy))
print("Shapiro A4:", stats.shapiro(a4_oxy))
print("Shapiro A5:", stats.shapiro(a5_oxy))

# Check equal variance across all 5 sections
print("Levene Oxygen:", stats.levene(a1_oxy, a2_oxy, a3_oxy, a4_oxy, a5_oxy))

# Normality & equal variance fail (p < 0.05) -> Run Kruskal-Wallis test
print("Kruskal-Wallis Oxygen:", stats.kruskal(a1_oxy, a2_oxy, a3_oxy, a4_oxy, a5_oxy))
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Post-hoc for Nitrate (follows ANOVA)
tukey = pairwise_tukeyhsd(endog=df['nitrate_mgL'], groups=df['river_section'], alpha=0.05)
print(tukey)
```
